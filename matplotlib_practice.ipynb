## Line Graphs:

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7,8,9]
y=[10,20,30,40,50,60,70,80,90]
plt.plot(x,y,color='k',label='Y not',linestyle='--',marker='o')
#color is to change the color of the 
#linestyle determines the type of the line like dashed line or a normal
#label is the title for the plot u have put
#Marker is the points u r tracing to be visible u use markers
plt.grid(True) #If u want a graph grid on yo plot 
plt.title('Idk man Jus practicing') #Title of the whole diagram
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis') #X and Y axis titles
plt.legend() #If u need a legend explainging shi
plt.show()
```

### Plotting Multiple line graphs can be done here

```python
#U can plot multiple lines on the same plot 
xx=[1,2,3,4,5,6,7,8,9]
yy=[10,20,30,40,50,60,70,80,90]
x=[1,2,3,4,5,6,7,8,9]
y=[8,16,24,32,40,48,56,64,72]
plt.plot(xx,yy,label='first one')
plt.plot(x,y,label='2nd one',color='g')
plt.legend()
plt.show()
```

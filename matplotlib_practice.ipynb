## Line Graphs:

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7,8,9]
y=[10,20,30,40,50,60,70,80,90]
plt.plot(x,y,color='k',label='Y not',linestyle='--',marker='o')
plt.style.use("fivethirtyeight")
#color is to change the color of the 
#linestyle determines the type of the line like dashed line or a normal
#label is the title for the plot u have put
#Marker is the points u r tracing to be visible u use markers
plt.grid(True) #If u want a graph grid on yo plot 
plt.title('Idk man Jus practicing') #Title of the whole diagram
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis') #X and Y axis titles
plt.legend() #If u need a legend explainging shi
plt.show()
```

### Plotting Multiple line graphs can be done here

```python
#U can plot multiple lines on the same plot 
xx=[1,2,3,4,5,6,7,8,9]
yy=[10,20,30,40,50,60,70,80,90]
x=[1,2,3,4,5,6,7,8,9]
y=[8,16,24,32,40,48,56,64,72]
plt.plot(xx,yy,label='first one')
plt.plot(x,y,label='2nd one',color='g')
plt.legend()
plt.show()
```

## Pie Charts

```python
slices=[60,40]  
color=['green','yellow'] #These values can be any values matplotlib will proportionalize it appropriatly
plt.pie(slices,labels=['sixty','forty'],wedgeprops={'edgecolor':'black'},colors=color)
plt.show()
```

### Using explode and shadow things 

```python
slices=[60,40]  
color=['green','yellow'] #These values can be any values matplotlib will proportionalize it appropriatly
explode=[0,0.1]
plt.pie(slices,labels=['sixty','forty'],wedgeprops={'edgecolor':'black'},colors=color,explode=explode,shadow=True,startangle=90,autopct='%1.2f%%')
#Start angle starts your pie chart from that angle
#autopct is a parameter that accepts a formatted string and used to display percentages
plt.show()
```

## Stack Plots

```python
years = [2018, 2019, 2020, 2021, 2022]
python_devs =     [300, 350, 400, 430, 470]
javascript_devs = [280, 290, 310, 330, 350]
java_devs =       [200, 210, 200, 195, 190]

plt.stackplot(years, python_devs, javascript_devs, java_devs,
              labels=["Python", "JavaScript", "Java"])
plt.xticks(rotation=90)
plt.legend(loc='upper left')
plt.show()
```

## Scatter Plots

```python
x=[1,2,3,4,5,6,7,8,9]
y=[8,16,24,32,40,48,56,64,72]
plt.scatter(x,y,s=100,marker='x',color='green',alpha=0.75,linewidths=1) #s is a parameter to set the size of the dots    
#alpha determines the opacity of the plot
#marker determines the type of point required
plt.show()
```

### U can have various colors for each point and various sizes too by passing an array of sizes and colors to the plot


```python
x = [1,2,3,4,5,6,7,8,9]
y = [8,16,24,32,40,48,56,64,72]
colors = [7,8,9,3,4,5,6,7,8]
sizes = [100,200,300,400,324,222,768,899,954]

sc = plt.scatter(x, y, s=sizes, marker='x', c=colors, alpha=0.75,
                 linewidths=1, cmap='Greens')

cbar = plt.colorbar(sc)
cbar.set_label('Idk man')
plt.show()
```
